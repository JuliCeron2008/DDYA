## 1) Descripción del problema

Una empresa necesita organizar los códigos numéricos de los productos almacenados en una bodega. Los códigos llegan en un orden aleatorio y se desea comparar el comportamiento de un **Árbol Binario de Búsqueda (BST)** frente a un **Árbol AVL** al insertar exactamente la misma secuencia de valores.

Adicionalmente, el sistema debe permitir recorrer ambos árboles mediante **Preorden**, **Inorden** y **Postorden**, y mostrar gráficamente su estructura (nodo raíz, hijos izquierdos, hijos derechos y conexiones) para que el usuario pueda comparar visualmente el resultado de cada estrategia.

**Datos de entrada:** el usuario indica cuántos códigos desea registrar (`n`) y luego ingresa cada código uno a uno. Con esos valores se construye primero el BST (sin reorganizar ni balancear) y luego el AVL (aplicando balanceo y rotaciones tras cada inserción), ambos insertando los códigos en el mismo orden de llegada.


## 2) Requerimientos

# Requerimientos funcionales:

# Punto 1: Árbol Binario de Búsqueda (BST)

RF 1: El programa solicita al usuario la cantidad `n` de códigos a registrar.

RF 2: El programa solicita, uno por uno, los `n` códigos y los almacena inicialmente en una lista (por ejemplo `codigos = [30, 20, 10, 25, 28, 27, 5, 50]`).

RF 3: El programa construye un BST insertando los códigos exactamente en el orden en que fueron ingresados, cumpliendo que en cada nodo los valores menores queden a la izquierda y los mayores a la derecha.

RF 4: El programa NO reorganiza ni balancea el BST bajo ninguna circunstancia.

RF 5: El programa muestra gráficamente el árbol BST obtenido (raíz, hijos y conexiones).

# Punto 2: Árbol AVL

RF 1: El programa construye un Árbol AVL utilizando los mismos códigos y el mismo orden de inserción que el BST.

RF 2: Después de cada inserción, el programa calcula el factor de balance de cada nodo afectado (`altura(izquierda) - altura(derecha)`).

RF 3: Si un nodo queda desbalanceado (factor mayor a 1 o menor a -1), el programa aplica la rotación correspondiente: simple a la derecha, simple a la izquierda, doble izquierda-derecha o doble derecha-izquierda.

RF 4: El programa muestra gráficamente el árbol AVL final, permitiendo compararlo visualmente con el BST.

# Punto 3, 4 y 5: Recorridos

RF 1: El programa realiza el recorrido Preorden (Raíz -> Izquierda -> Derecha) sobre ambos árboles y muestra el resultado.

RF 2: El programa realiza el recorrido Inorden (Izquierda -> Raíz -> Derecha) sobre ambos árboles y muestra el resultado, evidenciando que sobre un BST/AVL este recorrido siempre entrega los valores ordenados de menor a mayor.

RF 3: El programa realiza el recorrido Postorden (Izquierda -> Derecha -> Raíz) sobre ambos árboles y muestra el resultado.

RNF 1: La cantidad `n` de elementos debe ser indicada siempre por el usuario, validando que sea un entero positivo.

RNF 2: La inserción del BST debe implementarse manualmente, sin usar librerías externas de árboles.

RNF 3: El balanceo del AVL (cálculo de altura, factor de balance y rotaciones) debe implementarse manualmente.

RNF 4: Los tres recorridos (preorden, inorden, postorden) deben implementarse mediante recursividad.

RNF 5: La visualización de los árboles no debe limitarse a una lista de valores; debe mostrar gráficamente el nodo raíz, los hijos izquierdos, los hijos derechos y las conexiones entre nodos (se utiliza `matplotlib`).

RNF 6: Toda entrada por teclado debe validarse (tipo numérico, no vacía) sin que el programa se detenga por una excepción no controlada.

RNF 7: El código debe estar organizado en funciones y clases claras para facilitar su comprensión y mantenimiento.

**HU3.** *Como* analista de bodega, *quiero* recorrer ambos árboles en preorden, inorden y postorden, *para* analizar cómo cambia el orden de visita según la estructura de cada árbol.

## 7) Análisis de complejidad

- Justificación

El BST no se reorganiza, por lo que su altura depende completamente del orden de llegada de los códigos: en el mejor caso la secuencia de inserción produce un árbol lo más balanceado posible (altura ≈ `log₂(n)`); en el peor caso (secuencia ya ordenada ascendente o descendente) el árbol degenera en una estructura tipo lista enlazada de altura `n`. El AVL, en cambio, recalcula el factor de balance tras cada inserción y aplica rotaciones en O(1) cuando es necesario, lo que garantiza que su altura sea siempre `O(log n)`, sin importar el orden de llegada de los datos.

- Función T(n)

BST insertar (peor caso, árbol degenerado): `T(n) = T(n-1) + O(1)` → `T(n) = O(n)`.

AVL insertar (garantizado por el invariante de balance `|FB| ≤ 1`): `T(n) = T(n/2) + O(1)` → `T(n) = O(log n)`.

Recorridos (preorden, inorden, postorden) sobre cualquier árbol de `n` nodos: `T(n) = 2·T(n/2) + O(1)` en el caso balanceado, que se resuelve en `T(n) = O(n)` porque siempre se visita cada nodo exactamente una vez, sea cual sea la forma del árbol.

| Operación | Estructura | Mejor caso Ω | Peor caso O |
| :--- | :--- | :---: | :---: |
| Insertar | BST (sin balanceo) | Ω(1) — árbol vacío / Ω(log n) — inserción balanceada | O(n) — árbol degenerado |
| Buscar | BST (sin balanceo) | Ω(1) — valor en la raíz | O(n) — árbol degenerado |
| Insertar | AVL (balanceado) | Ω(1) — árbol vacío | O(log n) — garantizado por rotaciones |
| Buscar | AVL (balanceado) | Ω(1) — valor en la raíz | O(log n) — garantizado por balance |
| Rotación (LL/RR/LR/RL) | AVL | Ω(1) | O(1) |
| Preorden / Inorden / Postorden | BST y AVL | Ω(n) | O(n) |


## 8) Código funcional


```python
%pip install matplotlib
import sys
sys.setrecursionlimit(5000)
import matplotlib.pyplot as plt


def pedir_entero_positivo(mensaje):
    while True:
        try:
            texto = input(mensaje).strip()
            valor = int(texto)
            if valor >= 0:
                return valor
            print("Ingrese un número entero positivo.")
        except ValueError:
            print("Entrada inválida. Ingrese un número entero.")
        except (EOFError, KeyboardInterrupt):
            print("\nEntrada no disponible, se usará un valor por defecto.")
            return -1


def pedir_entero(mensaje):
    while True:
        try:
            texto = input(mensaje).strip()
            if texto == "":
                print("El campo no puede estar vacío.")
                continue
            return int(texto)
        except ValueError:
            print("Entrada inválida. Ingrese un número entero.")
        except (EOFError, KeyboardInterrupt):
            print("\nEntrada no disponible, se usará un valor por defecto.")
            return None


class NodoBST:
    def __init__(self, valor):
        self.valor = valor
        self.izquierdo = None
        self.derecho = None


class ArbolBinarioBusqueda:
    def __init__(self):
        self.raiz = None

    def esta_vacio(self):
        return self.raiz is None

    def insertar(self, valor):
        if self.raiz is None:
            self.raiz = NodoBST(valor)
            return
        actual = self.raiz
        while True:
            if valor < actual.valor:
                if actual.izquierdo is None:
                    actual.izquierdo = NodoBST(valor)
                    return
                actual = actual.izquierdo
            elif valor > actual.valor:
                if actual.derecho is None:
                    actual.derecho = NodoBST(valor)
                    return
                actual = actual.derecho
            else:
                return

    def buscar(self, valor):
        actual = self.raiz
        while actual is not None:
            if valor == actual.valor:
                return actual
            actual = actual.izquierdo if valor < actual.valor else actual.derecho
        return None

    def altura(self):
        return self._altura_rec(self.raiz)

    def _altura_rec(self, nodo):
        if nodo is None:
            return -1
        return 1 + max(self._altura_rec(nodo.izquierdo), self._altura_rec(nodo.derecho))

    def preorden(self):
        resultado = []
        self._preorden_rec(self.raiz, resultado)
        return resultado

    def _preorden_rec(self, nodo, resultado):
        if nodo is not None:
            resultado.append(nodo.valor)
            self._preorden_rec(nodo.izquierdo, resultado)
            self._preorden_rec(nodo.derecho, resultado)

    def inorden(self):
        resultado = []
        self._inorden_rec(self.raiz, resultado)
        return resultado

    def _inorden_rec(self, nodo, resultado):
        if nodo is not None:
            self._inorden_rec(nodo.izquierdo, resultado)
            resultado.append(nodo.valor)
            self._inorden_rec(nodo.derecho, resultado)

    def postorden(self):
        resultado = []
        self._postorden_rec(self.raiz, resultado)
        return resultado

    def _postorden_rec(self, nodo, resultado):
        if nodo is not None:
            self._postorden_rec(nodo.izquierdo, resultado)
            self._postorden_rec(nodo.derecho, resultado)
            resultado.append(nodo.valor)


class NodoAVL:
    def __init__(self, valor):
        self.valor = valor
        self.izquierdo = None
        self.derecho = None
        self.altura = 0


class ArbolAVL:
    def __init__(self):
        self.raiz = None

    def esta_vacio(self):
        return self.raiz is None

    def _altura(self, nodo):
        return nodo.altura if nodo else -1

    def _factor_balance(self, nodo):
        return self._altura(nodo.izquierdo) - self._altura(nodo.derecho) if nodo else 0

    def _actualizar_altura(self, nodo):
        nodo.altura = 1 + max(self._altura(nodo.izquierdo), self._altura(nodo.derecho))

    def _rotar_derecha(self, nodo):
        nueva_raiz = nodo.izquierdo
        nodo.izquierdo = nueva_raiz.derecho
        nueva_raiz.derecho = nodo
        self._actualizar_altura(nodo)
        self._actualizar_altura(nueva_raiz)
        return nueva_raiz

    def _rotar_izquierda(self, nodo):
        nueva_raiz = nodo.derecho
        nodo.derecho = nueva_raiz.izquierdo
        nueva_raiz.izquierdo = nodo
        self._actualizar_altura(nodo)
        self._actualizar_altura(nueva_raiz)
        return nueva_raiz

    def insertar(self, valor):
        self.raiz = self._insertar_rec(self.raiz, valor)

    def _insertar_rec(self, nodo, valor):
        if nodo is None:
            return NodoAVL(valor)
        if valor < nodo.valor:
            nodo.izquierdo = self._insertar_rec(nodo.izquierdo, valor)
        elif valor > nodo.valor:
            nodo.derecho = self._insertar_rec(nodo.derecho, valor)
        else:
            return nodo

        self._actualizar_altura(nodo)
        fb = self._factor_balance(nodo)

        if fb > 1 and valor < nodo.izquierdo.valor:
            return self._rotar_derecha(nodo)
        if fb < -1 and valor > nodo.derecho.valor:
            return self._rotar_izquierda(nodo)
        if fb > 1 and valor > nodo.izquierdo.valor:
            nodo.izquierdo = self._rotar_izquierda(nodo.izquierdo)
            return self._rotar_derecha(nodo)
        if fb < -1 and valor < nodo.derecho.valor:
            nodo.derecho = self._rotar_derecha(nodo.derecho)
            return self._rotar_izquierda(nodo)
        return nodo

    def buscar(self, valor):
        actual = self.raiz
        while actual is not None:
            if valor == actual.valor:
                return actual
            actual = actual.izquierdo if valor < actual.valor else actual.derecho
        return None

    def preorden(self):
        resultado = []
        self._preorden_rec(self.raiz, resultado)
        return resultado

    def _preorden_rec(self, nodo, resultado):
        if nodo is not None:
            resultado.append(nodo.valor)
            self._preorden_rec(nodo.izquierdo, resultado)
            self._preorden_rec(nodo.derecho, resultado)

    def inorden(self):
        resultado = []
        self._inorden_rec(self.raiz, resultado)
        return resultado

    def _inorden_rec(self, nodo, resultado):
        if nodo is not None:
            self._inorden_rec(nodo.izquierdo, resultado)
            resultado.append(nodo.valor)
            self._inorden_rec(nodo.derecho, resultado)

    def postorden(self):
        resultado = []
        self._postorden_rec(self.raiz, resultado)
        return resultado

    def _postorden_rec(self, nodo, resultado):
        if nodo is not None:
            self._postorden_rec(nodo.izquierdo, resultado)
            self._postorden_rec(nodo.derecho, resultado)
            resultado.append(nodo.valor)


def _obtener_posiciones(nodo, profundidad=0, contador=None, posiciones=None):
    if contador is None:
        contador = [0]
    if posiciones is None:
        posiciones = {}
    if nodo is None:
        return posiciones
    _obtener_posiciones(nodo.izquierdo, profundidad + 1, contador, posiciones)
    posiciones[id(nodo)] = (contador[0], -profundidad, nodo.valor)
    contador[0] += 1
    _obtener_posiciones(nodo.derecho, profundidad + 1, contador, posiciones)
    return posiciones


def dibujar_arbol(nodo, titulo="Árbol", color_nodo="#bfe1f0", color_raiz="#f4a259"):
    if nodo is None:
        print(f"{titulo}: árbol vacío, nada que dibujar.")
        return
    posiciones = _obtener_posiciones(nodo)
    ancho = max(4.5, len(posiciones) * 0.9)
    fig, ax = plt.subplots(figsize=(ancho, 5))

    def dibujar_conexiones(actual):
        if actual is None:
            return
        x1, y1, _ = posiciones[id(actual)]
        for hijo in (actual.izquierdo, actual.derecho):
            if hijo is not None:
                x2, y2, _ = posiciones[id(hijo)]
                ax.plot([x1, x2], [y1, y2], color="#7d8597", zorder=1, linewidth=1.5)
                dibujar_conexiones(hijo)

    dibujar_conexiones(nodo)

    for clave, (x, y, valor) in posiciones.items():
        color = color_raiz if clave == id(nodo) else color_nodo
        ax.scatter([x], [y], s=1100, color=color, edgecolors="#2b2d42", zorder=2, linewidths=1.5)
        ax.text(x, y, str(valor), ha="center", va="center", fontsize=11,
                fontweight="bold", color="#2b2d42", zorder=3)

    ax.set_title(titulo, fontsize=13, fontweight="bold")
    ax.axis("off")
    plt.tight_layout()
    plt.show()


def leer_codigos_usuario():
    n = pedir_entero_positivo("¿Cuántos códigos desea registrar?: ")
    if n <= 0:
        print("No se ingresaron códigos válidos; se usará el ejemplo del enunciado.")
        return [30, 20, 10, 25, 28, 27, 5, 50]
    codigos = []
    for i in range(1, n + 1):
        valor = pedir_entero(f"Ingrese el código {i}: ")
        if valor is None:
            break
        codigos.append(valor)
    return codigos if codigos else [30, 20, 10, 25, 28, 27, 5, 50]


try:
    codigos = leer_codigos_usuario()
except Exception as error:
    print(f"Ocurrió un problema leyendo los datos ({error}). Se usará el ejemplo del enunciado.")
    codigos = [30, 20, 10, 25, 28, 27, 5, 50]

print("Lista de códigos:", codigos)

bst = ArbolBinarioBusqueda()
for codigo in codigos:
    bst.insertar(codigo)

avl = ArbolAVL()
for codigo in codigos:
    avl.insertar(codigo)

dibujar_arbol(bst.raiz, "Árbol Binario de Búsqueda (BST)")
dibujar_arbol(avl.raiz, "Árbol AVL")

print("\n--- Recorridos BST ---")
print("Preorden :", bst.preorden())
print("Inorden  :", bst.inorden())
print("Postorden:", bst.postorden())
print("Altura BST:", bst.altura())

print("\n--- Recorridos AVL ---")
print("Preorden :", avl.preorden())
print("Inorden  :", avl.inorden())
print("Postorden:", avl.postorden())
print("Altura AVL:", avl.raiz.altura if avl.raiz else -1)

print("\n--- Comparación ---")
print(f"El BST no se balancea: su altura ({bst.altura()}) depende del orden de llegada.")
print(f"El AVL se balancea automáticamente: su altura ({avl.raiz.altura if avl.raiz else -1}) "
      f"se mantiene cercana a log2(n).")
print("El recorrido inorden de ambos árboles coincide y siempre entrega los códigos ordenados: "
      f"{bst.inorden() == avl.inorden()}")
```

## Pruebas de Desempeño

### 1. Medición empírica
Se construye, para cada tamaño `N`, un BST y un AVL insertando los códigos `0, 1, 2, ..., N-1` en orden ascendente (peor caso de inserción para el BST, ya que degenera en una lista). Sobre ambas estructuras se mide el tiempo de `buscar(N-1)` con `time.perf_counter()`, variando `N` desde 20 hasta 800 en pasos de 40.

### 2. Caso analizado
El valor buscado (`N-1`) es siempre el último insertado, por lo que en el BST degenerado la búsqueda recorre secuencialmente los `N` nodos (peor caso real de la estructura), mientras que en el AVL la búsqueda desciende como máximo `O(log N)` niveles gracias al balanceo.

### 3. Relación con la complejidad teórica
* **Teoría:** la búsqueda en el BST degenerado tiene un costo `T(N) = O(N)`; la búsqueda en el AVL tiene un costo `T(N) = O(log N)`.
* **Práctica:** la gráfica empírica muestra que el tiempo del BST crece con pendiente lineal a medida que aumenta `N`, mientras que la curva del AVL permanece casi plana, confirmando que el balanceo evita la degradación de rendimiento que sí sufre el BST sin balanceo.


```python
%pip install matplotlib

import time
import matplotlib.pyplot as plt


def bench(n_max, paso=40):
    print("Generando mediciones", end="")
    resultados_bst = []
    resultados_avl = []

    for size in range(20, n_max, paso):
        bst_b = ArbolBinarioBusqueda()
        avl_b = ArbolAVL()
        for i in range(size):
            bst_b.insertar(i)
            avl_b.insertar(i)

        ti = time.perf_counter()
        bst_b.buscar(size - 1)
        tf = time.perf_counter()
        resultados_bst.append((size, (tf - ti) * 1000))

        ti = time.perf_counter()
        avl_b.buscar(size - 1)
        tf = time.perf_counter()
        resultados_avl.append((size, (tf - ti) * 1000))

        print(".", end="")

    print("\nMediciones finalizadas.")
    return resultados_bst, resultados_avl


def dibujar_benchmark(bst_datos, avl_datos, dimensiones=(8, 5)):
    x_bst = [s for s, _ in bst_datos]
    y_bst = [t for _, t in bst_datos]
    x_avl = [s for s, _ in avl_datos]
    y_avl = [t for _, t in avl_datos]

    plt.figure(figsize=dimensiones)
    plt.plot(x_bst, y_bst, marker="o", color="#c0392b", label="BST degenerado - O(N)")
    plt.plot(x_avl, y_avl, marker="o", color="#1a5276", label="AVL balanceado - O(log N)")
    plt.title("Tiempo de búsqueda: BST degenerado vs AVL balanceado")
    plt.xlabel("Tamaño del árbol (N elementos)")
    plt.ylabel("Tiempo de ejecución (ms)")
    plt.grid(True)
    plt.legend()
    plt.show()


datos_bst, datos_avl = bench(800, paso=40)
dibujar_benchmark(datos_bst, datos_avl)
```
