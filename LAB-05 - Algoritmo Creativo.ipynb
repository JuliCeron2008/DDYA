## 1) Descripción del problema

Una tienda desea desarrollar un pequeño sistema para controlar la atención de clientes, registrar las operaciones realizadas y administrar su catálogo de productos.

Para resolver el problema se deberán utilizar tres estructuras de datos diferentes:

Cola: para administrar los clientes que esperan ser atendidos.
Pila: para guardar el historial de las últimas operaciones realizadas.
Lista doblemente enlazada: para administrar y recorrer el catálogo de productos.


## 2) Requerimientos:

# Requerimientos funcionales:

# Punto 1: Cola de clientes

RF 1: El programa recibe un numero n de clientes.

RF 2: Con base en el numero n de clientes el programa solicita el nombre de cada cliente.

RF 3: El programa organiza una cola de clientes dependiendo del orden de llegada. E.J: Ana -> Carlos -> Laura -> Mateo (Ana fue la primera en llegar y por ende la primera a la que se le registro el nombre).

RF 4: El programa deberá permitir agregar un nuevo cliente a la cola.

RF 5: El programa deberá permitir atender al siguiente cliente.

rF 6: El programa deberá permitir mostrar los clientes que siguen esperando.

# Punto 2: Pila de operaciones

RF 1: Cada vez que se atienda un cliente, el programa deberá guardar el nombre en una pila de clientes atendidos. E.J: | Laura  | <- Tope | Carlos | | Ana    | --------- (Laura fue la ultima en ser atendida).

RF 2: El programa deberá permitir agregar a la pila cada cliente que sea atendido.

RF 3: El programa deberá permitir mostrar el último cliente atendido.

RF 4: El programa deberá permitir deshacer la última atención. (El ultimo cliente que fue atendido sera retirado de la pila y debera regresar al inicio de la cola de clientes pendientes).

# Punto 3: Lista doblemente enlazada de productos

RF 1: El programa tendra una lista doblemente que tendra nodos que almacenaran(Código-Nombre-Precio) de los productos de la tienda.

RF 2: El programa deberá permitir agregar un producto al final de la lista.

RF 3: El programa deberá permitir mostrar todos los productos desde el primero hasta el último.

RF 4: El programa deberá permitir mostrar los productos desde el último hasta el primero.

RF 5: El programa deberá permitir buscar un producto utilizando su código.

RF 6: El programa deberá permitir eliminar un producto utilizando su código.

# Requerimientos no funcionales:

RNF 1: Las operaciones de cola y pila deben ejecutarse en tiempo O(1).

RNF 2. La Cola debe respetar el principio FIFO(First In, First Out, es decir: El primero en llegar es el primero en salir).

RNF 3. La Pila debe respetar el principio LIFO(Last In, First Out, es decir: El último en llegar es el primero en salir).

RNF 4. Cada nodo debe contener:

      anterior <- producto -> siguiente
      
      Esto nos permite recorrer el catálogo tanto hacia adelante como atrás.

RNF 5. El sistema debe funcionar mediante una interfaz sencilla, mostrando mensajes claros para que el usuario pueda introducir los datos y seleccionar las operaciones.

RNF 6. Los datos deben ser validados para evitar operaciones incorrectas, como códigos inválidos o precios que no sean numéricos.

RNF 7. El código debe estar organizado en funciones claras para facilitar su comprensión, modificación o mantenimiento.

RNF 8. Las operaciones deben implementarse aprovechando las características de cada estructura de datos y evitando recorridos innecesarios.







## 3) Historias de usuario

**HU1.** *Como* encargado de la tienda, *quiero* registrar a los clientes en el orden en que llegan, *para* garantizar una atención justa (primero en llegar, primero en ser atendido).
- *Criterio de aceptación:* al agregar clientes A, B, C, la cola queda en ese mismo orden; al atender, sale primero A.
 
**HU2.** *Como* encargado, *quiero* poder deshacer la última atención, *para* corregir errores sin perder el orden original del cliente.
- *Criterio de aceptación:* si el último atendido fue X, al deshacer, X sale de la pila y vuelve a ser el primero en la cola de espera.
 
**HU3.** *Como* encargado, *quiero* ver el catálogo completo en ambos sentidos, *para* revisar el inventario según necesite (por ejemplo del más reciente al más antiguo).
- *Criterio de aceptación:* recorrer la lista desde el primer nodo hasta None da el orden de inserción; recorrerla desde el último hasta None da el orden inverso.
 
**HU4.** *Como* encargado, *quiero* buscar un producto por su código, *para* consultar rápidamente su nombre y precio.
- *Criterio de aceptación:* dado un código existente, se retorna el producto correspondiente; dado uno inexistente, se informa que no se encontró.
 
**HU5.** *Como* encargado, *quiero* eliminar un producto del catálogo por su código, *para* mantener el inventario actualizado.
- *Criterio de aceptación:* al eliminar un código existente, el nodo se retira y los enlaces anterior/siguiente de sus vecinos quedan correctamente reconectados.

## 8)Codigo funcional

```python
def pedir_entero_positivo(mensaje):
    while True:
        try:
            valor = int(input(mensaje))
            if valor >= 0:
                return valor
            print("Ingrese un número positivo.")
        except ValueError:
            print("Entrada inválida. Ingrese un número entero.")


def pedir_flotante_positivo(mensaje):
    while True:
        try:
            valor = float(input(mensaje))
            if valor >= 0:
                return valor
            print("Ingrese un precio válido (mayor o igual a cero).")
        except ValueError:
            print("Entrada inválida. Ingrese un número decimal o entero.")


def pedir_texto(mensaje):
    while True:
        texto = input(mensaje).strip()
        if texto:
            return texto
        print("El campo no puede estar vacío.")


class ColaClientes:
    def __init__(self):
        self.cola = []

    def esta_vacia(self):
        return len(self.cola) == 0

    def encolar(self, nombre):
        self.cola.append(nombre)

    def desencolar(self):
        if not self.esta_vacia():
            return self.cola.pop(0)
        return None

    def reinsertar_al_inicio(self, nombre):
        self.cola.insert(0, nombre)

    def mostrar(self):
        if self.esta_vacia():
            print("Cola vacía.")
        else:
            print(" -> ".join(self.cola))


class PilaAtendidos:
    def __init__(self):
        self.pila = []

    def esta_vacia(self):
        return len(self.pila) == 0

    def apilar(self, nombre):
        self.pila.append(nombre)

    def desapilar(self):
        if not self.esta_vacia():
            return self.pila.pop()
        return None

    def ver_tope(self):
        if not self.esta_vacia():
            return self.pila[-1]
        return None

    def mostrar(self):
        if self.esta_vacia():
            print("Pila vacía.")
        else:
            for i in range(len(self.pila) - 1, -1, -1):
                if i == len(self.pila) - 1:
                    print(f"| {self.pila[i]} | <- Tope")
                else:
                    print(f"| {self.pila[i]} |")


class NodoProducto:
    def __init__(self, codigo, nombre, precio):
        self.codigo = codigo
        self.nombre = nombre
        self.precio = precio
        self.anterior = None
        self.siguiente = None


class ListaDobleProductos:
    def __init__(self):
        self.cabeza = None
        self.cola = None

    def esta_vacia(self):
        return self.cabeza is None

    def agregar_final(self, codigo, nombre, precio):
        nuevo = NodoProducto(codigo, nombre, precio)
        if self.esta_vacia():
            self.cabeza = nuevo
            self.cola = nuevo
        else:
            self.cola.siguiente = nuevo
            nuevo.anterior = self.cola
            self.cola = nuevo

    def mostrar_adelante(self):
        if self.esta_vacia():
            print("NULL")
            return
        actual = self.cabeza
        elementos = []
        while actual:
            elementos.append(actual.nombre)
            actual = actual.siguiente
        print("NULL <- " + " <-> ".join(elementos) + " -> NULL")

    def mostrar_atras(self):
        if self.esta_vacia():
            print("NULL")
            return
        actual = self.cola
        elementos = []
        while actual:
            elementos.append(actual.nombre)
            actual = actual.anterior
        print("NULL <- " + " <-> ".join(elementos) + " -> NULL")

    def buscar_por_codigo(self, codigo):
        actual = self.cabeza
        while actual:
            if str(actual.codigo) == str(codigo):
                return actual
            actual = actual.siguiente
        return None

    def eliminar_por_codigo(self, codigo):
        actual = self.buscar_por_codigo(codigo)
        if not actual:
            print("Producto no encontrado.")
            return False

        if actual == self.cabeza and actual == self.cola:
            self.cabeza = None
            self.cola = None
        elif actual == self.cabeza:
            self.cabeza = actual.siguiente
            self.cabeza.anterior = None
        elif actual == self.cola:
            self.cola = actual.anterior
            self.cola.siguiente = None
        else:
            actual.anterior.siguiente = actual.siguiente
            actual.siguiente.anterior = actual.anterior
        return True


cola = ColaClientes()
pila = PilaAtendidos()
catalogo = ListaDobleProductos()

catalogo.agregar_final(101, "Arroz", 4500)
catalogo.agregar_final(102, "Leche", 3800)
catalogo.agregar_final(103, "Pan", 2500)
catalogo.agregar_final(104, "Café", 12000)

n = pedir_entero_positivo("¿Cuántos clientes desea registrar?: ")
for i in range(1, n + 1):
    nom = pedir_texto(f"Ingrese el nombre del cliente {i}: ")
    cola.encolar(nom)

print("Cola inicial:")
cola.mostrar()

while True:
    print("\n1. Agregar cliente a la cola")
    print("2. Atender siguiente cliente")
    print("3. Mostrar clientes esperando")
    print("4. Mostrar último atendido")
    print("5. Deshacer última atención")
    print("6. Agregar producto al final")
    print("7. Mostrar catálogo (inicio a fin)")
    print("8. Mostrar catálogo (fin a inicio)")
    print("9. Buscar producto por código")
    print("10. Eliminar producto por código")
    print("0. Salir")

    opcion = input("Opción: ").strip()

    if opcion == "1":
        nom = pedir_texto("Ingrese el nuevo cliente: ")
        cola.encolar(nom)
        cola.mostrar()
    elif opcion == "2":
        cliente = cola.desencolar()
        if cliente:
            pila.apilar(cliente)
            print(f"Atendiendo a: {cliente}")
            print("Clientes esperando: ", end="")
            cola.mostrar()
        else:
            print("No hay clientes.")
    elif opcion == "3":
        cola.mostrar()
    elif opcion == "4":
        pila.mostrar()
    elif opcion == "5":
        deshecho = pila.desapilar()
        if deshecho:
            cola.reinsertar_al_inicio(deshecho)
            print(f"Se devolvió a la cola: {deshecho}")
            cola.mostrar()
        else:
            print("No hay atenciones para deshacer.")
    elif opcion == "6":
        cod = pedir_entero_positivo("Código: ")
        nom = pedir_texto("Nombre: ")
        prec = pedir_flotante_positivo("Precio: ")
        catalogo.agregar_final(cod, nom, prec)
    elif opcion == "7":
        catalogo.mostrar_adelante()
    elif opcion == "8":
        catalogo.mostrar_atras()
    elif opcion == "9":
        cod = pedir_texto("Código a buscar: ")
        prod = catalogo.buscar_por_codigo(cod)
        if prod:
            print(f"{prod.codigo} - {prod.nombre} - ${prod.precio}")
        else:
            print("No encontrado.")
    elif opcion == "10":
        cod = pedir_texto("Código a eliminar: ")
        catalogo.eliminar_por_codigo(cod)
        catalogo.mostrar_adelante()
    elif opcion == "0":
        break
    else:
        print("Opción inválida.")
```

## Pruebas de enunciado

Se validan las operaciones descritas en el enunciado:
1. **Cola (FIFO):** Orden de llegada, atención del primero y llegada de nuevo cliente al final.
2. **Pila (LIFO) y Rollback:** Apilado de atendidos, consulta de tope y reversión de atención a la cabeza de la cola.
3. **Lista doblemente enlazada:** Búsqueda por código 103, eliminación del nodo y reconexión de punteros.

| Caso de prueba | Entrada | Salida esperada | Salida obtenida | Estado |
| :--- | :--- | :--- | :--- | :---: |
| Cola inicial | `["Ana", "Carlos", "Laura", "Mateo"]` | `["Ana", "Carlos", "Laura", "Mateo"]` | `["Ana", "Carlos", "Laura", "Mateo"]` | Pasa |
| Atender primer cliente | `cola.desencolar()` | `"Ana"` | `"Ana"` | Pasa |
| Cola tras atención | Estado interno de cola | `["Carlos", "Laura", "Mateo"]` | `["Carlos", "Laura", "Mateo"]` | Pasa |
| Llegada nuevo cliente | `cola.encolar("Sofía")` | `["Carlos", "Laura", "Mateo", "Sofía"]` | `["Carlos", "Laura", "Mateo", "Sofía"]` | Pasa |
| Tope de atendidos | `pila.ver_tope()` tras atender 3 | `"Laura"` | `"Laura"` | Pasa |
| Desapilar último atendido | `pila.desapilar()` | `"Laura"` | `"Laura"` | Pasa |
| Nuevo tope tras desapilar | `pila.ver_tope()` | `"Carlos"` | `"Carlos"` | Pasa |
| Reversión a la cola | `cola.reinsertar_al_inicio("Laura")` | `"Laura"` en posición 0 | `"Laura"` en posición 0 | Pasa |
| Buscar código 103 | `catalogo.buscar_por_codigo(103)` | `"Pan"` | `"Pan"` | Pasa |
| Eliminar código 103 | `catalogo.eliminar_por_codigo(103)` | `True` | `True` | Pasa |
| Comprobar eliminación | `catalogo.buscar_por_codigo(103)` | `None` | `None` | Pasa |
| Integridad de enlaces | Recorrido tras borrado | `["Arroz", "Leche", "Café"]` | `["Arroz", "Leche", "Café"]` | Pasa |

```python
c = ColaClientes()
for nombre in ["Ana", "Carlos", "Laura", "Mateo"]:
    c.encolar(nombre)

test("Cola inicial", list(c.cola), ["Ana", "Carlos", "Laura", "Mateo"])

atendido = c.desencolar()
test("Atender primer cliente", atendido, "Ana")
test("Cola tras atención", list(c.cola), ["Carlos", "Laura", "Mateo"])

c.encolar("Sofía")
test("Llegada nuevo cliente", list(c.cola), ["Carlos", "Laura", "Mateo", "Sofía"])

p = PilaAtendidos()
p.apilar("Ana")
p.apilar("Carlos")
p.apilar("Laura")

test("Tope de atendidos", p.ver_tope(), "Laura")

ultimo_deshecho = p.desapilar()
test("Desapilar último atendido", ultimo_deshecho, "Laura")
test("Nuevo tope tras desapilar", p.ver_tope(), "Carlos")

c.reinsertar_al_inicio(ultimo_deshecho)
test("Reversión a la cola", c.cola[0], "Laura")

catalogo = ListaDobleProductos()
catalogo.agregar_final(101, "Arroz", 4500)
catalogo.agregar_final(102, "Leche", 3800)
catalogo.agregar_final(103, "Pan", 2500)
catalogo.agregar_final(104, "Café", 12000)

prod_buscado = catalogo.buscar_por_codigo(103)
test("Buscar código 103", prod_buscado.nombre if prod_buscado else None, "Pan")

eliminado = catalogo.eliminar_por_codigo(103)
test("Eliminar código 103", eliminado, True)

prod_eliminado = catalogo.buscar_por_codigo(103)
test("Comprobar eliminación", prod_eliminado, None)

nombres_catalogo = []
actual = catalogo.cabeza
while actual:
    nombres_catalogo.append(actual.nombre)
    actual = actual.siguiente

test("Integridad de enlaces", nombres_catalogo, ["Arroz", "Leche", "Café"])
```

## Pruebas de Desempeño

### 1. Medición empírica
Se evalúa el tiempo de ejecución (en ms) de la operación `buscar_por_codigo(codigo)` aumentando el tamaño del catálogo $N$ desde 10 hasta 1000 productos con pasos de 50, usando `time.perf_counter()`.

### 2. Caso analizado
Se consulta en cada prueba el último elemento de la lista ($N - 1$). Al ser una lista doblemente enlazada sin acceso por índice, el algoritmo realiza un recorrido secuencial obligatorio desde la cabeza hasta el final (peor caso).

### 3. Relación con la complejidad teórica
* **Teoría:** La búsqueda lineal tiene un costo temporal en el peor caso de $T(N) = O(N)$.
* **Práctica:** La gráfica empírica muestra una curva con pendiente lineal positiva, confirmando que a mayor cantidad de productos registrados, el tiempo de búsqueda aumenta de forma directamente proporcional.

```python
%pip install matplotlib

import time
import matplotlib.pyplot as plt

def bench(n: int, step: int = 50) -> list[tuple[int, float]]:
    print("Generando mediciones", end="")
    results: list[tuple[int, float]] = []

    for size in range(10, n, step):
        lista = ListaDobleProductos()
        for i in range(size):
            lista.agregar_final(i, f"Prod_{i}", 1000 + i)

        ti = time.perf_counter()
        _ = lista.buscar_por_codigo(size - 1)
        tf = time.perf_counter()

        delta = (tf - ti) * 1000
        print(".", end="")
        results.append((size, delta))

    print("\nMediciones finalizadas.")
    return results

def dibujar(x, y, dimensiones=(7, 5)):
    assert len(x) == len(y)
    plt.figure(figsize=dimensiones)
    plt.plot(x, y, marker='o', color='b', label='Búsqueda O(N)')
    plt.title('Tiempo de Ejecución: Búsqueda en Lista Doblemente Enlazada')
    plt.xlabel('Tamaño del catálogo (N elementos)')
    plt.ylabel('Tiempo de ejecución (ms)')
    plt.grid(True)
    plt.legend()
    plt.show()

tiempos = bench(1000, step=50)

x = [size for size, timing in tiempos]
y = [timing for size, timing in tiempos]

dibujar(x, y)
```
