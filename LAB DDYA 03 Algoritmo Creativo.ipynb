## Problema - Organización y búsqueda de productos en una tienda


## 1) Descripción del problema

El objetivo de este programa es gestionar el inventario de códigos numéricos de productos de una tienda. Como los productos se registran a medida que llegan, la lista inicial de códigos se encuentra desordenada.

El sistema debe resolver dos tareas principales mediante algoritmos clásicos:

1. Ordenamiento de datos: Reorganizar los n códigos de productos en orden ascendente (de menor a mayor) utilizando el algoritmo Merge Sort (Dividir y Conquistar).

2. Búsqueda eficiente: Permitir al usuario consultar la existencia de un código específico dentro de la lista previamente ordenada utilizando el algoritmo de Búsqueda Binaria.

Entrada: Un entero positivo n, seguido de n códigos numéricos desordenados y un código objetivo a buscar.

Salida: La lista de códigos organizada de menor a mayor y un mensaje indicando si el código buscado fue encontrado o no.

## 8) Código funcional 


```python
def merge_sort(codigos, inicio, fin):
    """
    Ordena la lista de códigos numéricos usando Merge Sort (Dividir y Conquistar).
    """
    if inicio < fin:
        medio = (inicio + fin) // 2

        # División recursiva
        merge_sort(codigos, inicio, medio)
        merge_sort(codigos, medio + 1, fin)

        # Creación de sublistas temporales
        izquierda = codigos[inicio:medio + 1]
        derecha = codigos[medio + 1:fin + 1]

        i = 0
        j = 0
        k = inicio

        # Combinación ordenada
        while i < len(izquierda) and j < len(derecha):
            if izquierda[i] <= derecha[j]:
                codigos[k] = izquierda[i]
                i += 1
            else:
                codigos[k] = derecha[j]
                j += 1
            k += 1

        # Elementos restantes de la izquierda
        while i < len(izquierda):
            codigos[k] = izquierda[i]
            i += 1
            k += 1

        # Elementos restantes de la derecha
        while j < len(derecha):
            codigos[k] = derecha[j]
            j += 1
            k += 1


def busqueda_binaria(codigos, codigo_buscado):
    """
    Busca un código en la lista ordenada usando Búsqueda Binaria.
    Retorna la posición (índice) si existe, o -1 si no fue encontrado.
    """
    inicio = 0
    fin = len(codigos) - 1

    while inicio <= fin:
        medio = (inicio + fin) // 2

        if codigos[medio] == codigo_buscado:
            return medio
        elif codigo_buscado < codigos[medio]:
            fin = medio - 1
        else:
            inicio = medio + 1

    return -1


def pedir_entero(mensaje):
    """
    Solicita una entrada y valida que sea estrictamente un número entero.
    Evita caídas por letras, símbolos o números decimales.
    """
    while True:
        entrada = input(mensaje).strip()
        try:
            valor = int(entrada)
            return valor
        except ValueError:
            print(" -> Entrada no válida. Debe ingresar un número entero (sin letras ni decimales).")


def pedir_cantidad_productos():
    """
    Valida que la cantidad de productos sea un número entero mayor a cero.
    """
    while True:
        cantidad = pedir_entero("Ingrese la cantidad de productos: ")
        if cantidad > 0:
            return cantidad
        print(" -> La cantidad de productos debe ser un número mayor a 0.")


def main():
    print("=" * 45)
    print("  SISTEMA DE GESTIÓN Y BÚSQUEDA DE INVENTARIO")
    print("=" * 45)

    # 1. Ingreso de cantidad y códigos con validación
    cantidad = pedir_cantidad_productos()
    codigos = []

    for i in range(cantidad):
        codigo = pedir_entero(f"Ingrese el código del producto [{i + 1}/{cantidad}]: ")
        codigos.append(codigo)

    # 2. Ordenamiento con Merge Sort
    print("\n" + "-" * 45)
    print("Códigos originales ingresados:", codigos)
    merge_sort(codigos, 0, len(codigos) - 1)
    print("Códigos ordenados (Merge Sort):", codigos)
    print("-" * 45)

    # 3. Consulta con Búsqueda Binaria
    codigo_buscado = pedir_entero("\nIngrese el código que desea buscar: ")
    posicion = busqueda_binaria(codigos, codigo_buscado)

    # 4. Entrega de resultados
    print("\n--- RESULTADO DE LA CONSULTA ---")
    if posicion != -1:
        print(f"✓ El código {codigo_buscado} fue encontrado.")
        print(f"✓ Posición en la lista ordenada (índice): {posicion}")
    else:
        print(f"✗ El código {codigo_buscado} no fue encontrado en el inventario.")


if __name__ == "__main__":
    main()
```
