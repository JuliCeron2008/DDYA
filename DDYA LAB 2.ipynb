## Enunciado

Ejercicio: Organización de precios en una tienda

Una tienda necesita organizar los precios de sus productos. La cantidad de productos puede variar, por lo que el usuario deberá indicar cuántos precios desea registrar.

El programa debe solicitar:

1. La cantidad de productos, representada por n.

2. El precio de cada uno de los productos.

3. El tipo de ordenamiento que se desea mostrar.

Después de ingresar los datos, el programa deberá utilizar el algoritmo Insertion Sort para organizar los precios.

Punto 1. Orden ascendente: Ordenar los precios desde el menor hasta el mayor, con el fin de mostrar primero los productos más económicos.

Punto 2. Orden descendente: Ordenar los precios desde el mayor hasta el menor, con el fin de mostrar primero los productos más costosos.

# Especificacion



¿Qué recibe el programa? (Entradas)

Un número entero n que indica la cantidad de productos a registrar (n > 0).

Una secuencia de n números que representan los precios de cada producto.

Un número entero (1 o 2) que define el tipo de ordenamiento:

1: Orden ascendente (menor a mayor).

2: Orden descendente (mayor a menor).

¿Qué debe producir el programa? (Salida)

La lista con los n precios organizados según el criterio seleccionado.

Restricciones:n debe ser un número entero positivo.Los precios ingresados deben ser valores numéricos válidos mayores o iguales a cero.

El ordenamiento se debe realizar obligatoriamente mediante Insertion Sort.


# Diseño

Estrategia general de solución:

Usaremos la estrategia incremental mediante el algoritmo Insertion Sort (ordenamiento por inserción).

Tipo de algoritmo e implementación:

Tipo de algoritmo: Incremental.

Implementación: Iterativa (usando ciclos for y while).

Invariante de ciclo:Al inicio de cada iteración i del ciclo principal, la sublista izquierda precios[0 ... i-1] siempre contiene los elementos ya procesados y ordenados.

Explicación de la lógica:

1. Tomamos el eleemnto actual en la posición i y lo guardamos en una variable llamada actual.

2. Comparamos actual con los elementos de su izquierda uno por uno.

3. Si elegimos Ascendente: Movemos a la derecha los elementos que sean mayores que actuall.

4. Si elegimos Descendente: Movemos a la derecha los elementos que sean menores que actual.

5.Colocamos actual en el hueco que quedó libre.

# Codigo

```python
def insertion_sort(precios, orden):
    n = len(precios)
    for i in range(1, n):
        actual = precios[i]
        j = i - 1

        if orden == 1:
            while j >= 0 and precios[j] > actual:
                precios[j + 1] = precios[j]
                j = j - 1
        else:
            while j >= 0 and precios[j] < actual:
                precios[j + 1] = precios[j]
                j = j - 1

        precios[j + 1] = actual

    return precios


def pedirPrecios(n):
    precios = []
    for i in range(n):
        precio = float(input("Ingrese el precio del producto " + str(i + 1) + ": "))
        precios.append(precio)
    return precios


def main():
    
    n = int(input("Ingrese la cantidad de productos: "))
    precios = pedirPrecios(n)

    print("1. Orden ascendente")
    print("2. Orden descendente")
    opcion = int(input("Seleccione el tipo de ordenamiento: "))

    resultado = insertion_sort(precios, opcion)

    print("Precios organizados:", resultado)


main()
```
